# Q. Statistical Mechanics of the Quantum Oscillator

**Full Worked Solution**

## 1. The Partition Function

The canonical partition function is

$$Z(T)=\sum_{j=0}^{\infty}e^{-\beta E_j},\qquad \beta=\frac{1}{k_BT},\qquad E_j=\hbar\omega\left(j+\frac12\right)$$

Substituting $E_j$:

$$Z(T)=\sum_{j=0}^{\infty}e^{-\beta\hbar\omega\left(j+\frac12\right)}=e^{-\beta\hbar\omega/2}\sum_{j=0}^{\infty}e^{-\beta\hbar\omega j}=e^{-\beta\hbar\omega/2}\sum_{j=0}^{\infty}\left(e^{-\beta\hbar\omega}\right)^{j}$$

The sum is a geometric series with common ratio $r=e^{-\beta\hbar\omega}$. Since $\beta,\hbar,\omega>0$, we have $0<r<1$, so the series converges to $\dfrac{1}{1-r}$. Hence

$$\boxed{Z(T)=\frac{e^{-\hbar\omega/(2k_BT)}}{1-e^{-\hbar\omega/(k_BT)}}}$$


## 2. Thermodynamic Potentials

It is convenient to first write $\ln Z$:

$$\ln Z=-\frac{\beta\hbar\omega}{2}-\ln\left(1-e^{-\beta\hbar\omega}\right)$$

### Helmholtz Free Energy

$$F(T)=-k_BT\ln Z=-k_BT\left[-\frac{\beta\hbar\omega}{2}-\ln\left(1-e^{-\beta\hbar\omega}\right)\right]$$

Since $k_BT\beta=1$, the first term simplifies to $\dfrac{\hbar\omega}{2}$:

$$\boxed{F(T)=\frac{\hbar\omega}{2}+k_BT\ln\left(1-e^{-\hbar\omega/(k_BT)}\right)}$$

### Internal Energy

$$U(T)=-\frac{\partial \ln Z}{\partial \beta}=-\frac{\partial}{\partial\beta}\left[-\frac{\beta\hbar\omega}{2}-\ln\left(1-e^{-\beta\hbar\omega}\right)\right]$$

$$=\frac{\hbar\omega}{2}+\frac{\partial}{\partial\beta}\ln\left(1-e^{-\beta\hbar\omega}\right)=\frac{\hbar\omega}{2}+\frac{\hbar\omega\,e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}$$

Multiplying numerator and denominator by $e^{\beta\hbar\omega}$:

$$\boxed{U(T)=\frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^{\hbar\omega/(k_BT)}-1}}$$

This is the familiar zero-point energy plus the mean occupation number (Bose-Einstein factor) times $\hbar\omega$.


## 3. Entropy

Let $x=\dfrac{\hbar\omega}{k_BT}$, so that $F(T)=\dfrac{\hbar\omega}{2}+k_BT\ln\left(1-e^{-x}\right)$.

$$S=-\left(\frac{\partial F}{\partial T}\right)=-k_B\ln\left(1-e^{-x}\right)-k_BT\,\frac{\partial}{\partial T}\ln\left(1-e^{-x}\right)$$

Using $\dfrac{dx}{dT}=-\dfrac{x}{T}$ and $\dfrac{d}{dx}\ln(1-e^{-x})=\dfrac{e^{-x}}{1-e^{-x}}$:

$$\frac{\partial}{\partial T}\ln(1-e^{-x})=\frac{e^{-x}}{1-e^{-x}}\left(-\frac{x}{T}\right)$$

so that

$$S=-k_B\ln(1-e^{-x})+k_B\,\frac{x\,e^{-x}}{1-e^{-x}}$$

$$\boxed{S(T)=k_B\left[\frac{x}{e^{x}-1}-\ln\left(1-e^{-x}\right)\right],\qquad x=\frac{\hbar\omega}{k_BT}}$$

### Verification that $U=F+TS$

$$F+TS=\frac{\hbar\omega}{2}+k_BT\ln(1-e^{-x})+T\,k_B\left[\frac{x}{e^{x}-1}-\ln(1-e^{-x})\right]$$

The $\ln(1-e^{-x})$ terms cancel, leaving

$$F+TS=\frac{\hbar\omega}{2}+\frac{k_BT\,x}{e^{x}-1}$$

Since $k_BT\,x=k_BT\cdot\dfrac{\hbar\omega}{k_BT}=\hbar\omega$,

$$F+TS=\frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^{x}-1}=U(T)\qquad\checkmark$$

confirming consistency with the earlier expression for $U(T)$.


## 4. Heat Capacity and Hyperbolic Form

Starting from $U(T)=\dfrac{\hbar\omega}{2}+\dfrac{\hbar\omega}{e^{x}-1}$ with $x=\dfrac{\hbar\omega}{k_BT}$:

$$C_V=\frac{\partial U}{\partial T}=\hbar\omega\,\frac{\partial}{\partial T}\left(e^{x}-1\right)^{-1}=-\hbar\omega\,\frac{e^{x}}{(e^{x}-1)^2}\,\frac{dx}{dT}$$

Using $\dfrac{dx}{dT}=-\dfrac{x}{T}$:

$$C_V=\frac{\hbar\omega\,x\,e^{x}}{T\,(e^{x}-1)^2}$$

Since $\hbar\omega=xk_BT$, we get $\dfrac{\hbar\omega\,x}{T}=k_Bx^2$, so

$$C_V=k_B\,x^2\,\frac{e^{x}}{(e^{x}-1)^2}$$

### Converting to hyperbolic form

Factor $e^{x}-1=e^{x/2}\left(e^{x/2}-e^{-x/2}\right)=2e^{x/2}\sinh\!\left(\frac{x}{2}\right)$, using $\sinh(u)=\dfrac{e^{u}-e^{-u}}{2}$. Then

$$(e^{x}-1)^2=4e^{x}\sinh^2\!\left(\frac{x}{2}\right)$$

so

$$\frac{e^{x}}{(e^{x}-1)^2}=\frac{e^{x}}{4e^{x}\sinh^2(x/2)}=\frac{1}{4\sinh^2(x/2)}$$

Substituting back:

$$C_V=k_B\,x^2\cdot\frac{1}{4\sinh^2(x/2)}=k_B\left(\frac{x}{2}\right)^2\frac{1}{\sinh^2(x/2)}$$

Recalling $\dfrac{x}{2}=\dfrac{\hbar\omega}{2k_BT}$:

$$\boxed{C_V=k_B\left(\frac{\hbar\omega}{2k_BT}\right)^2\frac{1}{\sinh^2\!\left(\dfrac{\hbar\omega}{2k_BT}\right)}}$$

which is exactly the required hyperbolic form.


## 5. Asymptotic Behavior: High-Temperature Limit

In the high-temperature limit $k_BT\gg\hbar\omega$, we have $x=\dfrac{\hbar\omega}{k_BT}\to 0$, so $\dfrac{x}{2}\to 0$ as well.

For small arguments, $\sinh(u)\approx u+\dfrac{u^3}{6}+\dots\approx u$, so $\sinh^2(x/2)\approx (x/2)^2$. Substituting into the heat capacity:

$$C_V=k_B\left(\frac{x}{2}\right)^2\frac{1}{\sinh^2(x/2)}\;\xrightarrow[x\to0]{}\;k_B\left(\frac{x}{2}\right)^2\cdot\frac{1}{(x/2)^2}=k_B$$

**Result:** In the high-temperature limit, $C_V\to k_B$, independent of $\omega$, $\hbar$, or $T$.

**Comparison with classical equipartition:** The classical equipartition theorem assigns $\tfrac12 k_B$ of energy to each quadratic degree of freedom in the Hamiltonian. A harmonic oscillator has two such quadratic terms - kinetic energy $\tfrac{p^2}{2m}$ and potential energy $\tfrac12 m\omega^2x^2$ - giving a classical energy $U_{classical}=k_BT$ and heat capacity $C_V=k_B$.

This matches exactly the high-temperature limit obtained from the quantum result. This is expected on physical grounds: when $k_BT\gg\hbar\omega$, the thermal energy is much larger than the spacing between quantized energy levels, so the discreteness of the spectrum becomes negligible and the system behaves classically (the correspondence principle). At low temperatures ($k_BT\ll\hbar\omega$), by contrast, $C_V\to0$ exponentially - a purely quantum effect (energy-level "freeze-out") with no classical analogue.


## Numerical Verification

As a sanity check, the cell below evaluates the derived closed-form expressions for $U$, $F$, $S$, and $C_V$ and compares $Z$ against a brute-force sum over energy levels, confirms $U=F+TS$, and confirms the high-temperature limit $C_V\to k_B$.

```python
import numpy as np

hbar = 1.0545718e-34   # J s
kB   = 1.380649e-23    # J / K
omega = 5.0e13         # rad/s, arbitrary test value

def brute_force_Z(T, n_terms=20000):
    beta = 1.0 / (kB * T)
    j = np.arange(n_terms)
    E = hbar * omega * (j + 0.5)
    return np.sum(np.exp(-beta * E))

def Z_closed_form(T):
    beta = 1.0 / (kB * T)
    x = beta * hbar * omega
    return np.exp(-x/2) / (1 - np.exp(-x))

def U_closed_form(T):
    x = hbar*omega/(kB*T)
    return hbar*omega/2 + hbar*omega/(np.exp(x)-1)

def F_closed_form(T):
    x = hbar*omega/(kB*T)
    return hbar*omega/2 + kB*T*np.log(1 - np.exp(-x))

def S_closed_form(T):
    x = hbar*omega/(kB*T)
    return kB*( x/(np.exp(x)-1) - np.log(1-np.exp(-x)) )

def Cv_closed_form(T):
    x = hbar*omega/(kB*T)
    return kB * (x/2)**2 / (np.sinh(x/2))**2

# --- check Z against brute-force sum ---
for T in [50, 300, 5000]:
    print(f"T = {T:>6} K   Z_closed = {Z_closed_form(T):.6e}   Z_brute = {brute_force_Z(T):.6e}")

print()

# --- check U = F + T*S ---
for T in [50, 300, 5000]:
    U = U_closed_form(T)
    F = F_closed_form(T)
    S = S_closed_form(T)
    print(f"T = {T:>6} K   U = {U:.4e} J   F + T*S = {F + T*S:.4e} J   (match: {np.isclose(U, F+T*S)})")

print()

# --- check high-T limit of Cv ---
for T in [1e2, 1e4, 1e7, 1e10]:
    print(f"T = {T:.0e} K   C_V = {Cv_closed_form(T):.6e} J/K   k_B = {kB:.6e} J/K   ratio C_V/k_B = {Cv_closed_form(T)/kB:.6f}")
```
