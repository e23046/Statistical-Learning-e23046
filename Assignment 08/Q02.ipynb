# Maximum Entropy Distribution under Mean and Variance Constraints

**Problem.** Let $X \in (-\infty,\infty)$ be a continuous random variable with density $p(x)$. Given only:

$$\mathbb{E}[X] = \mu, \qquad \mathbb{E}[(X-\mu)^2] = \sigma^2,$$

find the density $p(x)$ that maximizes the differential entropy

$$H[p] = -\int_{-\infty}^{\infty} p(x)\ln p(x)\, dx.$$

## 1. Setting up the constrained optimization

We must maximize $H[p]$ subject to three constraints (normalization plus the two given moment conditions):

$$
\begin{aligned}
C_0[p] &= \int p(x)\,dx - 1 = 0, \\
C_1[p] &= \int x\,p(x)\,dx - \mu = 0, \\
C_2[p] &= \int (x-\mu)^2\,p(x)\,dx - \sigma^2 = 0.
\end{aligned}
$$

This is a constrained functional optimization problem, solved using **Lagrange multipliers** $\lambda_0,\lambda_1,\lambda_2$. We form the Lagrangian functional

$$
\mathcal{L}[p] = -\int p(x)\ln p(x)\,dx \;-\; \lambda_0\left(\int p(x)\,dx - 1\right) \;-\; \lambda_1\left(\int x\,p(x)\,dx - \mu\right) \;-\; \lambda_2\left(\int (x-\mu)^2 p(x)\,dx - \sigma^2\right).
$$

## 2. Functional derivative (calculus of variations)

Treat $\mathcal L$ as a functional of $p(x)$, i.e. $\mathcal{L} = \int \ell\big(x,p(x)\big)\,dx$ with

$$\ell(x,p) = -p\ln p - \lambda_0 p - \lambda_1 x\,p - \lambda_2 (x-\mu)^2 p + \text{(constant terms independent of }p).$$

Since there is no derivative of $p$ inside the integrand, the Euler-Lagrange condition reduces to the simple stationarity condition $\dfrac{\partial \ell}{\partial p} = 0$ for every $x$:

$$
\frac{\partial \ell}{\partial p} = -\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x-\mu)^2 = 0.
$$

Solving for $p(x)$:

$$
\boxed{p(x) = \exp\Big(-1-\lambda_0-\lambda_1 x - \lambda_2 (x-\mu)^2\Big)}
$$

This confirms that the maximum-entropy density, subject only to mean and variance information, **must belong to the exponential family with a linear-plus-quadratic exponent** -- i.e. it is necessarily Gaussian in *form*. We now pin down $\lambda_0,\lambda_1,\lambda_2$ using the constraints themselves.

## 3. Completing the square: fixing $\lambda_1$

Group the exponent as a quadratic in $x$:

$$
-\lambda_1 x - \lambda_2(x-\mu)^2 = -\lambda_2\left(x - \mu - \frac{\lambda_1}{2\lambda_2}\right)^2 + \Big(\text{terms independent of } x\Big).
$$

So, absorbing constants into a single normalization prefactor $A = e^{-1-\lambda_0+(\text{const.})}$,

$$
p(x) = A\, \exp\!\left[-\lambda_2\left(x - \left(\mu + \frac{\lambda_1}{2\lambda_2}\right)\right)^2\right].
$$

This is a Gaussian centered at $\mu + \dfrac{\lambda_1}{2\lambda_2}$. But the mean constraint $\mathbb{E}[X]=\mu$ requires the center of this (symmetric, unimodal) density to equal $\mu$ exactly. Hence

$$
\mu + \frac{\lambda_1}{2\lambda_2} = \mu \quad\Longrightarrow\quad \lambda_1 = 0.
$$

With $\lambda_1 = 0$, the density simplifies to

$$
p(x) = A\, e^{-\lambda_2 (x-\mu)^2}.
$$

## 4. Fixing $A$ and $\lambda_2$ from the constraints

**Normalization.** Using the Gaussian integral $\displaystyle\int_{-\infty}^{\infty} e^{-\lambda_2 u^2}\,du = \sqrt{\pi/\lambda_2}$ (for $\lambda_2>0$, required for integrability):

$$
\int_{-\infty}^{\infty} p(x)\,dx = A\sqrt{\frac{\pi}{\lambda_2}} = 1 \quad\Longrightarrow\quad A = \sqrt{\frac{\lambda_2}{\pi}}.
$$

**Variance constraint.** With $u = x-\mu$, and using $\displaystyle\int_{-\infty}^{\infty} u^2 e^{-\lambda_2 u^2}\,du = \frac{1}{2\lambda_2}\sqrt{\frac{\pi}{\lambda_2}}$:

$$
\mathbb{E}[(X-\mu)^2] = A\int_{-\infty}^{\infty} u^2 e^{-\lambda_2 u^2}\,du = \sqrt{\frac{\lambda_2}{\pi}}\cdot \frac{1}{2\lambda_2}\sqrt{\frac{\pi}{\lambda_2}} = \frac{1}{2\lambda_2} = \sigma^2.
$$

So

$$
\lambda_2 = \frac{1}{2\sigma^2}, \qquad A = \sqrt{\frac{1}{2\pi\sigma^2}}.
$$

Since the mean constraint is automatically satisfied by symmetry once $\lambda_1=0$ (the density is symmetric about $\mu$), all three constraints are now satisfied simultaneously.

## 5. Result

Substituting back:

$$
\boxed{p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)}
$$

which is exactly the density of $X \sim \mathcal{N}(\mu, \sigma^2)$, as required to show.

**Interpretation.** Among *all* probability densities on $(-\infty,\infty)$ with a specified mean $\mu$ and variance $\sigma^2$ (and no other information assumed), the Gaussian is the *unique* least-biased (maximum entropy) choice. Any other distribution with the same mean and variance encodes additional, unjustified structure (skewness, multimodality, heavier/lighter tails, etc.) that is not supported by the given constraints, and therefore has strictly lower entropy.

**Second-order check (this is indeed a maximum, not a saddle point).** The entropy functional $H[p] = -\int p\ln p\,dx$ is strictly concave in $p$ (since $-p\ln p$ is a strictly concave function of $p$ for $p>0$), and the constraint set (fixed mean and variance) is an affine/convex subspace of densities. A strictly concave functional maximized over a convex constraint set has a unique global maximizer, so the stationary point found above is guaranteed to be the global entropy maximizer, not merely a critical point.

## 6. Numerical verification

Below we verify symbolically/numerically that:
1. $p(x)$ integrates to 1,
2. $\mathbb{E}[X] = \mu$,
3. $\mathbb{E}[(X-\mu)^2] = \sigma^2$,
4. among a family of alternative densities with the *same* mean and variance, the Gaussian has the *highest* entropy.

```python
import sympy as sp

x = sp.symbols('x', real=True)
sigma = sp.symbols('sigma', positive=True)
mu = sp.symbols('mu', real=True)

p = sp.exp(-(x-mu)**2/(2*sigma**2)) / sp.sqrt(2*sp.pi*sigma**2)

# 1. Normalization
norm = sp.integrate(p, (x, -sp.oo, sp.oo))
print('Integral of p(x) dx  =', sp.simplify(norm))

# 2. Mean
mean = sp.integrate(x*p, (x, -sp.oo, sp.oo))
print('E[X]                 =', sp.simplify(mean))

# 3. Variance
var = sp.integrate((x-mu)**2*p, (x, -sp.oo, sp.oo))
print('E[(X-mu)^2]           =', sp.simplify(var))
```

```python
import numpy as np
from scipy import integrate, stats

# Numerical sanity check with concrete values
mu_val, sigma_val = 2.0, 1.5

def gaussian_pdf(x):
    return stats.norm.pdf(x, loc=mu_val, scale=sigma_val)

norm_check, _ = integrate.quad(gaussian_pdf, -np.inf, np.inf)
mean_check, _ = integrate.quad(lambda x: x*gaussian_pdf(x), -np.inf, np.inf)
var_check, _  = integrate.quad(lambda x: (x-mu_val)**2*gaussian_pdf(x), -np.inf, np.inf)

print(f'Normalization : {norm_check:.6f}  (target 1)')
print(f'Mean          : {mean_check:.6f}  (target {mu_val})')
print(f'Variance      : {var_check:.6f}  (target {sigma_val**2})')
```

```python
# Compare entropy of the Gaussian against several OTHER densities
# that share the same mean and variance, to confirm the Gaussian has
# the highest (maximum) differential entropy among them.

def differential_entropy(pdf, lo=-30, hi=30):
    integrand = lambda x: -pdf(x)*np.log(pdf(x)) if pdf(x) > 0 else 0.0
    val, _ = integrate.quad(integrand, lo, hi, limit=200)
    return val

# Candidate 1: the MaxEnt Gaussian itself
H_gauss = differential_entropy(gaussian_pdf)

# Candidate 2: Logistic distribution rescaled to match mean & variance
# Logistic(loc, scale) has variance = (pi^2/3) * scale^2
scale_logistic = sigma_val / (np.pi/np.sqrt(3))
logistic_pdf = lambda x: stats.logistic.pdf(x, loc=mu_val, scale=scale_logistic)
H_logistic = differential_entropy(logistic_pdf)

# Candidate 3: Laplace distribution rescaled to match mean & variance
# Laplace(loc, b) has variance = 2*b^2
b_laplace = sigma_val/np.sqrt(2)
laplace_pdf = lambda x: stats.laplace.pdf(x, loc=mu_val, scale=b_laplace)
H_laplace = differential_entropy(laplace_pdf)

# Candidate 4: Uniform distribution matched in mean & variance
# Uniform on [a,b]: mean=(a+b)/2, var=(b-a)^2/12
half_width = sigma_val*np.sqrt(3)
a_u, b_u = mu_val-half_width, mu_val+half_width
uniform_pdf = lambda x: stats.uniform.pdf(x, loc=a_u, scale=b_u-a_u)
H_uniform = differential_entropy(uniform_pdf, lo=a_u-1, hi=b_u+1)

print(f'Entropy (Gaussian) : {H_gauss:.5f}   <-- should be the LARGEST')
print(f'Entropy (Logistic) : {H_logistic:.5f}')
print(f'Entropy (Laplace)  : {H_laplace:.5f}')
print(f'Entropy (Uniform)  : {H_uniform:.5f}')

assert H_gauss > H_logistic
assert H_gauss > H_laplace
assert H_gauss > H_uniform
print()
print('Confirmed: the Gaussian attains strictly higher entropy than every')
print('other tested density sharing the same mean and variance.')
```

## 7. Conclusion

Starting from the entropy functional and imposing only the mean and variance as constraints via Lagrange multipliers, the calculus-of-variations stationarity condition forces the density to have the exponential-quadratic form $p(x)=A e^{-\lambda_1 x - \lambda_2(x-\mu)^2}$. Matching this form to the constraints (symmetry about $\mu$, normalization, and the prescribed variance $\sigma^2$) uniquely fixes

$$
\lambda_1 = 0,\qquad \lambda_2=\frac{1}{2\sigma^2},\qquad A=\frac{1}{\sqrt{2\pi\sigma^2}},
$$

giving

$$
p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right) = \mathcal{N}(\mu,\sigma^2),
$$

i.e. the **Normal distribution is the maximum entropy distribution on $(-\infty,\infty)$ subject to a fixed mean and variance**, as required. $\blacksquare$
